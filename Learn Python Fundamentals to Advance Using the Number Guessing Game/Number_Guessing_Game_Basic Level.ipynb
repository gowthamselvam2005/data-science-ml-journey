```python
import random


secret_number=random.randint(1,100)
for i in range (1,8):
    user_input=int(input("Enter the Guessing Number:"))

    if user_input > 1 or 100 < user_input:
        print("Enter the Valid Number between 1 -100")
    elif user_input > secret_number:
        print("Your Number Is Lesser")
    elif user_input < secret_number:
        print("Your Number Is Higher")
    else:
        print("Vaalthukal Neenga Sariya Kandu Pudichitingaa..!")
        break
if secret_number == user_input:
    print("Congradualtion Your Guessing Number is Correct")
```
