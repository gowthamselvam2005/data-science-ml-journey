# Data Structure 

### List

#### Append Function in list

```python
n=list(map(int,input().split()))
l1=[]
for i in n:
       if i%2==0:
         l1.append(i)
print(*l1)
```

#### Pop the particular Index Value

```python
l2=[10,20,30,45,55,34,26]
l2.pop(0)
print(*l2)
```

#### Count the Particular Number & Alphabets from the List 

```python
l2=[10,20,30,45,55,34,26]
print(l2.count(10))
```

```python
l3=["aa","a","ab","as","bb","er","aa","fg"]
print(l3.count("aa"))
```

```python
l3=l2.copy()
print(l3)
```

# Set Functions

```python
s = {10, 20}
s.add(30)
print(s)
```

```python
s = {10, 20, 30}
s.remove(20)
print(s)
```

```python
s = {10, 20, 30}
s.discard(50)
print(s)
```

```python
s = {10, 20, 30}
print(s.pop())
print(s)
```

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.union(b))
```

```python
a = {1, 2, 3}
b = {2, 3, 4}
print(a.intersection(b))
```

```python
a = {1, 2, 3}
b = {2, 3, 4}
print(a.difference(b))
```

### Dictionary Functions

```python
student = {"name": "Gowtham", "age": 20}
print(student.get("name"))
```

```python
student = {"name": "Gowtham", "age": 20}
print(student.keys())
print(student.values())
print(student.items())
student.update({"age": 20})
print(student)
student.pop("age")
print(student)
```

##  Data Frame Using Dictionary

```python
import pandas as pd

data = {
    "Name": ["Gowtham", "Shruthi", "Parthiban"],
    "Age": [20, 21, 22],
    "City": ["Salem", "Namakkal", "Cudalore"]
}

df = pd.DataFrame(data)

print(df)
df.to_csv("students.csv", index=False)

print("CSV file created successfully!")
```
